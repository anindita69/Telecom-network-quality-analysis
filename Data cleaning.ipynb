```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv("January_MyCall_2022.csv")
```

```python
df.columns
```

```python
df.info()
```

```python
# Trimming white spaces from text columns
text_cols = ["operator", "inout_travelling", "network_type", "calldrop_category", "state_name"]

for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
df["rating"].nunique()
```

```python
(df["latitude"] == -1).sum()
```

```python
(df["longitude"] == -1).sum()
```

```python
(df["state_name"] == "NA").sum()
```

```python
df["latitude"] = df["latitude"].replace(-1, np.nan)
df["longitude"] = df["longitude"].replace(-1, np.nan)
df["state_name"] = df["state_name"].replace("NA", "Not Available")
```

```python
assert df["rating"].between(1, 5).all(), "Found a value outside 1 - 5"
```
